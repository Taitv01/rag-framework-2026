## 3. Basic RAG Setup

Let's create a simple RAG system using local embeddings (no API key needed for embeddings):

```python
# Initialize RAG
# Note: For LLM, you'll need OPENAI_API_KEY in .env file
rag = NaiveRAG(
    llm_provider='openai',
    llm_model='gpt-4o-mini',
    embedding_provider='huggingface',  # Local embeddings, no API key needed
    vector_store_provider='faiss',
    chunk_size=500,
    chunk_overlap=50,
)

print('✅ RAG initialized!')
```

## 4. Adding Documents

Add documents to the knowledge base:

```python
# Sample documents
documents = [
    """Python is a high-level, general-purpose programming language. 
    Its design philosophy emphasizes code readability with the use of 
    significant indentation. Python is dynamically typed and garbage-collected.""",
    
    """Python was created by Guido van Rossum and was first released in 1991. 
    The language has evolved significantly over the years, with Python 3.0 
    being released in 2008.""",
    
    """Python has a comprehensive standard library and a large ecosystem of 
    third-party packages. The Python Package Index (PyPI) hosts over 400,000 
    packages for various tasks.""",
    
    """Python is widely used in data science and machine learning due to 
    libraries like NumPy, Pandas, Scikit-learn, TensorFlow, and PyTorch.""",
]

# Add documents
num_chunks = rag.add_texts(documents)
print(f'✅ Added {num_chunks} chunks from {len(documents)} documents')
```

## 5. Querying

Now let's query the knowledge base:

```python
# Query
question = 'What is Python?'
answer = rag.query(question)

print(f'Question: {question}')
print(f'Answer: {answer}')
```

```python
# Query with sources
result = rag.query_with_sources('Who created Python?')

print(f"Answer: {result['answer']}")
print(f"\nSources:")
for i, source in enumerate(result['sources'], 1):
    print(f"  [{i}] {source['content'][:100]}...")
```

## 6. Document Retrieval

You can also retrieve documents without generating an answer:

```python
# Retrieve relevant documents
docs = rag.retrieve('machine learning', k=3)

print(f'Retrieved {len(docs)} documents:')
for i, doc in enumerate(docs, 1):
    print(f"\n[{i}] {doc.page_content[:150]}...")
```

## 7. Advanced RAG (Optional)

For better results, use AdvancedRAG with hybrid search and re-ranking:

```python
# Advanced RAG with hybrid search
advanced_rag = AdvancedRAG(
    llm_provider='openai',
    llm_model='gpt-4o-mini',
    use_hybrid=True,      # Vector + BM25
    use_reranking=True,    # Cross-encoder re-ranking
)

# Add documents
advanced_rag.add_texts(documents)

# Query with detailed results
result = advanced_rag.query_detailed('What is Python used for?')

print(f"Answer: {result['answer']}")
print(f"\nTransformed query: {result['transformed_query']}")
print(f"Docs retrieved: {result['total_docs_retrieved']}")
print(f"Relevant docs: {result['relevant_docs_count']}")
```

## 9. Evaluation

Evaluate your RAG system:

```python
from src.evaluation import RAGEvaluator

# Create evaluator
evaluator = RAGEvaluator(rag.llm)

# Test data
test_data = [
    {'question': 'What is Python?', 'expected': 'Python is a programming language'},
    {'question': 'Who created Python?', 'expected': 'Guido van Rossum'},
]

# Evaluate
def query_func(question):
    result = rag.query_with_sources(question)
    return {
        'answer': result['answer'],
        'contexts': [s['content'] for s in result['sources']]
    }

report = evaluator.evaluate(query_func, test_data)

# Print report
evaluator.print_report(report)
```
